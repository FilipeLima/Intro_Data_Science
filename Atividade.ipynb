```python
salario_bruto = float(input("Informe o valor do Salário Bruto: "))

imposto_renda = salario_bruto * 0.11
previdencia_social = salario_bruto * 0.08
sindicato = salario_bruto * 0.01

descontos_totais = imposto_renda + previdencia_social + sindicato

salario_liquido = salario_bruto - descontos_totais

print(f"Imposto de Renda: R$ {imposto_renda:.2f}")
print(f"Previdência Social: R$ {previdencia_social:.2f}")
print(f"Sindicato: R$ {sindicato:.2f}")
print(f"Descontos Totais: R$ {descontos_totais:.2f}")
print(f"Salário Líquido: R$ {salario_liquido:.2f}")
```

#### 6 - Crie linhas de código em que o usuário deve informar apenas os números do CPF (que contém 11 caracteres). Nessas linhas de código você precisa criar um alerta de erro caso o usuário preencha os caracteres "." ou "-". Depois disso crie um código que mostre na tela o CPF no formato tradicional (xxx.xxx.xxx-xx). Resumindo: O usuário precisa informar apenas os números do CPF(xxxxxxxxxxx), mas, no momento da exibição na tela o CPF informado deve estar no formato xxx.xxx.xxx-xx.

```python
cpf = input("Informe apenas os números do CPF (11 dígitos, sem pontos ou traços): ")

if "." in cpf or "-" in cpf:
    print("Erro! Não informe pontos (.) ou traços (-). Digite apenas os números.")
elif len(cpf) != 11 or not cpf.isdigit():
    print("Erro! O CPF deve conter exatamente 11 números.")
else:
    cpf_formatado = f"{cpf[0:3]}.{cpf[3:6]}.{cpf[6:9]}-{cpf[9:11]}"
    print(f"CPF formatado: {cpf_formatado}")
```

7 - Crie linhas de código em que o usuário deve informar apenas os números do CPF (que contém 11 caracteres). Neste caso o usuário poderá informar o CPF no formato que quiser, contanto que ao final haja uma codificação de CPF válida (11 caracteres numéricos).

Resumindo: O usuário pode informar o CPF no formato que desejar, mas, no momento da exibição na tela o CPF informado deve estar no formato xxx.xxx.xxx-xx. Caso o que o usuário informar não resulte no formato padrão dos CPF (xxx.xxx.xxx-xx) uma mensagem de erro deverá ser informada na tela.

```python
cpf = input("Informe o CPF (em qualquer formato): ")

cpf_numeros = "".join(caractere for caractere in cpf if caractere.isdigit())

if len(cpf_numeros) != 11:
    print("Erro! O CPF informado não é válido. É necessário conter 11 números.")
else:
    cpf_formatado = f"{cpf_numeros[0:3]}.{cpf_numeros[3:6]}.{cpf_numeros[6:9]}-{cpf_numeros[9:11]}"
    print(f"CPF formatado: {cpf_formatado}")
```

Nessas linhas de código o usuário poderá optar em informar ou não o primeiro dígito obrigatório ("9"), mas, ao final, seu programa deve sempre informar na tela o número do celular no formato padrão 9xxxx-xxxx.

A partir dessas informações seu sistema deve apurar o "Resultado Financeiro" (que é o resultado da diferença entre o valor do "Salário" e o valor das "Despesas") e também apurar a "Situação Financeira" (que será considerada "Boa", quando o Resultado Financeiro for maior ou igual a zero, e "Ruim", quando o Resultado Financeiro for negativo).

Ao final o seu sistema deve mostrar na tela as seguintes informações:

Salário: R$\$$ 5.520,45

Despesas: R$ 3.895,86

Resultado Financeiro: R$\$$ 1.624,59

Situação Financeira: Boa

```python
salario_texto = input("Informe o valor do Salário (ex: 5.520,45): ")
despesas_texto = input("Informe o valor das Despesas (ex: 3.895,86): ")

salario = float(salario_texto.replace(".", "").replace(",", "."))
despesas = float(despesas_texto.replace(".", "").replace(",", "."))

resultado_financeiro = salario - despesas

if resultado_financeiro >= 0:
    situacao_financeira = "Boa"
else:
    situacao_financeira = "Ruim"

salario_br = f"{salario:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".")
despesas_br = f"{despesas:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".")
resultado_br = f"{resultado_financeiro:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".")

print(f"Salário: R$ {salario_br}")
print(f"Despesas: R$ {despesas_br}")
print(f"Resultado Financeiro: R$ {resultado_br}")
print(f"Situação Financeira: {situacao_financeira}")
```

Digamos que você trabalha num setor que atende o público por meio de processos administrativos. As listas representam os meses e as quantidades de processos atendidos, respectivamente:

meses = ["janeiro", "fevereiro", "março", "abril", "maio", "junho", "julho", "agosto", "setembro", "outubro", "novembro", "dezembro"]

atendimentos_semestre_1 = [350, 280, 420, 560, 325, 295]
atendimentos_semestre_2 = [490, 550, 620, 516, 380, 478]

Crie um sistema onde seja possível evidenciar na tela o meses em que ocorreram as maiores e menores quantidades de atendimentos, bem como evidenciar o percentual que esses atendimentos representam em relação ao total de atendimentos no ano.

No sistema deve ser evidenciada as seguintes menagens:


"O mês de xxx foi o que registrou a maior quantidade de atendimentos no ano com yyy atendimentos, o que representa zz % do total de atendimentos no ano."

"O mês de xxx foi o que registrou a menor quantidade de atendimentos no ano com yyy atendimentos, o que representa apenas zz % do total de atendimentos no ano."

```python
meses = ["janeiro", "fevereiro", "março", "abril", "maio", "junho", "julho", "agosto", "setembro", "outubro", "novembro", "dezembro"]
atendimentos_semestre_1 = [350, 280, 420, 560, 325, 295]
atendimentos_semestre_2 = [490, 550, 620, 516, 380, 478]

atendimentos_ano = atendimentos_semestre_1 + atendimentos_semestre_2

total_atendimentos = sum(atendimentos_ano)

maior_quantidade = max(atendimentos_ano)
menor_quantidade = min(atendimentos_ano)

posicao_maior = atendimentos_ano.index(maior_quantidade)
posicao_menor = atendimentos_ano.index(menor_quantidade)

mes_maior = meses[posicao_maior]
mes_menor = meses[posicao_menor]

percentual_maior = (maior_quantidade / total_atendimentos) * 100
percentual_menor = (menor_quantidade / total_atendimentos) * 100

print(f"O mês de {mes_maior} foi o que registrou a maior quantidade de atendimentos no ano com {maior_quantidade} atendimentos, o que representa {percentual_maior:.2f}% do total de atendimentos no ano.")
print(f"O mês de {mes_menor} foi o que registrou a menor quantidade de atendimentos no ano com {menor_quantidade} atendimentos, o que representa apenas {percentual_menor:.2f}% do total de atendimentos no ano.")
```
